## Problem Description

The purpose of this analysis is to determine whether Alloy B has a significantly higher mean tensile strength than Alloy A at a significance level of α = 0.05.

The following hypotheses were tested:

$$
H_0: μB- μA=0
$$

$$
H_0: μB- μA>0
$$

where μA represents the population mean tensile strength of Alloy A and μB represents the population mean tensile strength of Alloy B.

Because the samples consist of separate specimens from two independent groups and the population standard deviations are unknown, a two-sample t-test was used.

## Methodology

Python was used to calculate descriptive statistics and perform a right-tailed two-sample t-test. Each alloy consisted of 10 independent tensile-strength measurements. The sample mean and sample standard deviation were calculated for each alloy.

A pooled two-sample t-test was used to compare the means. The null hypothesis assumes no difference in mean tensile strength between the alloys, while the alternative hypothesis states that Alloy B has a higher mean tensile strength. The significance level was set at α = 0.05.

The decision rule was based on the p-value. If the p-value was less than 0.05, the null hypothesis was rejected.

```python
# Step 2: Import required modules and tensile strength data
import numpy as np
from scipy import stats

# Tensile strength data in MPa
alloy_A = np.array([
    310, 315, 305, 312, 318,
    310, 309, 314, 315, 308
])

alloy_B = np.array([
    318, 320, 325, 315, 322,
    319, 317, 321, 324, 320
])

# Step 3: Sample statistics
n_A = len(alloy_A)
n_B = len(alloy_B)

mean_A = np.mean(alloy_A)
mean_B = np.mean(alloy_B)

std_A = np.std(alloy_A, ddof=1)
std_B = np.std(alloy_B, ddof=1)

var_A = np.var(alloy_A, ddof=1)
var_B = np.var(alloy_B, ddof=1)

print("Alloy A mean:", round(mean_A, 3))
print("Alloy B mean:", round(mean_B, 3))
print("Alloy A standard deviation:", round(std_A, 3))
print("Alloy B standard deviation:", round(std_B, 3))

# Step 4: Pooled variance and pooled standard deviation
pooled_variance = (
    ((n_A - 1) * var_A + (n_B - 1) * var_B)
    / (n_A + n_B - 2)
)

pooled_std = np.sqrt(pooled_variance)

# Step 5: Test statistic
standard_error = pooled_std * np.sqrt(
    1 / n_A + 1 / n_B
)

t_stat = (mean_B - mean_A) / standard_error

df = n_A + n_B - 2

# Step 6: One-tailed p-value
p_value = 1 - stats.t.cdf(t_stat, df)

print("\nMean difference B - A:", round(mean_B - mean_A, 3))
print("Pooled standard deviation:", round(pooled_std, 3))
print("t statistic:", round(t_stat, 3))
print("Degrees of freedom:", df)
print("One-tailed p-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Reject H0")
    print("Alloy B has significantly higher mean tensile strength.")
else:
    print("Fail to reject H0")
    print("There is not sufficient evidence that Alloy B is stronger.")
```
