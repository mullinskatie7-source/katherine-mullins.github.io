## Problem Description

The purpose of this assignment is to analyze a dataset that is expected to follow an approximately normal distribution. Factory part diameter measurements will be analyzed using Python to determine the mean, median, mode, range, variance, and standard deviation.

Graphical methods will also be used to examine the distribution of the measurements. A stem-and-leaf plot, box plot, and histogram will be created to display the distribution, spread, and central tendency of the factory part diameters.

## Methodology

The factory part diameter dataset was imported into Google Colab using the Python Pandas library. The diameter measurements were selected from the Diameter_mm column of the dataset. Python statistical functions were then used to calculate the mean, median, mode, range, variance, and standard deviation of the 200 diameter measurements.

```python
# Step 2: Import required modules and define constants
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the CSV file
df = pd.read_csv("factory_part_diameters.csv")

# Display the first few rows
print(df.head())

# Step 3: Select the factory part diameter measurements
data = df["Diameter_mm"].dropna()

print(f"Number of measurements: {len(data)}")

# Step 4: Calculate basic statistics

mean = data.mean()
median = data.median()
mode = data.mode().iloc[0]
data_range = data.max() - data.min()
variance = data.var(ddof=1)
standard_deviation = data.std(ddof=1)

print(f"Mean: {mean:.4f} mm")
print(f"Median: {median:.4f} mm")
print(f"Mode: {mode:.4f} mm")
print(f"Range: {data_range:.4f} mm")
print(f"Variance: {variance:.6f} mm²")
print(f"Standard Deviation: {standard_deviation:.4f} mm")

# Step 5: Create a Stem-and-Leaf Plot

sorted_data = sorted(data)

stems = {}

for value in sorted_data:
    value_str = f"{value:.3f}"
    stem = value_str[:-1]
    leaf = value_str[-1]

    if stem not in stems:
        stems[stem] = []

    stems[stem].append(leaf)

print("Stem-and-Leaf Plot")
print("Stem | Leaves")

for stem, leaves in stems.items():
    print(f"{stem} | {' '.join(leaves)}")

# Step 6: Create a Box Plot

plt.figure(figsize=(8, 4))

plt.boxplot(data, vert=False)

plt.title("Box Plot of Factory Part Diameters")
plt.xlabel("Diameter (mm)")

plt.grid(axis="x", alpha=0.3)

plt.show()
# Step 7: Create a Histogram

plt.figure(figsize=(8, 5))

plt.hist(data, bins=10, edgecolor="black")

plt.title("Histogram of Factory Part Diameters")
plt.xlabel("Diameter (mm)")
plt.ylabel("Frequency")

plt.grid(axis="y", alpha=0.3)

plt.show()
```
