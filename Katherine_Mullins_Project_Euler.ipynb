## Problem Description

Project Euler Problem 1 asks for the sum of all natural numbers below 1000 that are multiples of 3 or 5. The problem gives a smaller example first: the multiples of 3 or 5 below 10 are 3, 5, 6, and 9, and their sum is 23. The task is to apply the same idea to every natural number below 1000.

In this problem, a number should be included in the sum if it is divisible by 3 or divisible by 5. Numbers that are divisible by both, such as 15, should only be counted once. The goal is to develop an algorithm that checks each number below 1000, identifies the values that meet the divisibility rule, and adds them together to produce the final answer.

## Methodology

The algorithm solves the problem by checking each natural number below 1000. A `for` loop is used to examine every integer from 1 through 999. For each number, the modulo operator `%` checks whether the number is evenly divisible by 3 or 5. If the remainder is zero for either divisor, the number is added to a running total.

This approach prevents numbers that are multiples of both 3 and 5, such as 15, from being counted twice because each number is checked once inside a single loop. After all numbers below 1000 are tested, the final value of the running total is printed as the solution. This method is simple, easy to verify, and directly follows the logic of the problem statement.


```python
# Step 2: Project Euler Problem 1
# Multiples of 3 or 5

# This variable stores the running total
total = 0

# Check every natural number below 1000
for number in range(1, 1000):

    # If the number is divisible by 3 or 5, add it to the total
    if number % 3 == 0 or number % 5 == 0:
        total += number

# Print the final answer
print("The sum of all multiples of 3 or 5 below 1000 is:", total)
```
