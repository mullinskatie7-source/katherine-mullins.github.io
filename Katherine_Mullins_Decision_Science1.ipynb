## Problem Description

Imagine you've inspected 12 optical assemblies with 3/12 assemblies needing rework. We want to determine how many units passed, how many require rework, and the overall rework rate.

## Methodology

Python provides a useful function called len(), which counts the number of elements in a list.

An important statement is .count() which counts how many times a specific value appears.

A conditional like if result =="rework": checks whetehr the current results require rework. Changing the condition to if result =="pass": then it checks whetehr the current results are passing. This is part of a for loop that repeats a block of code for each item in the list. The if statement determines which condition is true and what code to execute.

The main formula being used is the empirical formula where we estimate the rate of rework based on the data collected. p-hat is the estimated rework, x is the number of times an outcome occurs, and n is the total number of observatioins.
$$
\hat{p}= \frac{x}{n}
$$


```python
# --------------------------------
# Step 1: Inspection Dataset
# --------------------------------

inspection_results = [
    "pass",
    "pass",
    "rework",
    "pass",
    "pass",
    "pass",
    "rework",
    "pass",
    "pass",
    "pass",
    "pass",
    "rework"
]

# --------------------------------
# Step 2: Count Inspection Results
# --------------------------------
total_inspected = len(inspection_results)
rework_count = inspection_results.count("rework")
pass_count = inspection_results.count("pass")

# --------------------------------
# Step 3: Display Results
# --------------------------------
print("Total inspected:", total_inspected)
print("Rework count:", rework_count)
print("Pass count:", pass_count)

#----------------------------------
# Step 4: Calculate Inspection Rates
#----------------------------------
rework_rate = (rework_count / total_inspected) * 100
pass_rate = (pass_count / total_inspected) * 100

#----------------------------------
# Step 5: Display Results
#----------------------------------
print("Rework Rate:", rework_rate, "%")
print("Pass Rate:", pass_rate, "%")

```
